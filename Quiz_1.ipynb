**10. Plot the image.**
This is a tricky one because the image (as you see above) is huge. In fact, it is too big to fit in Google Colab's memory. So...read a `rasterio.windows.Window` of the image. See the documentation. I suggest a window not larger than 3000 x 2000. BUT, you can use AI for this question. You can use `matplotlib` or `ee` and `geemap`.

````python
from rasterio.windows import Window
import matplotlib.pyplot as plt
import numpy as np
````

Read a subset (window = 3000 * 2000)

````python
w, h = 3000, 2000

# Compute offsets so the window is taken from the center of the raster
# max(..., 0) ensures offsets are not negative if raster is smaller
col_off = max((img.width - w)//2, 0)
row_off = max((img.height - h)//2, 0)

# Create a rasterio window using the calculated offsets and size
win = Window(col_off, row_off, w, h)

# Read only this window (subset) from the raster to avoid memory issues
subset = img.read(window=win)              # shape: (4 bands, h, w)

# Take the first three bands and rearrange axes to (rows, cols, bands) for plotting
rgb = subset[:3].transpose(1, 2, 0)        # shape: (h, w, 3)

# Plot the RGB image
plt.figure(figsize=(10,6))
plt.imshow(rgb)
plt.title("Center window (RGB)")
plt.axis("off")
plt.show()

# Print min and max values to verify the window contains real data
print("min/max in window (band 1):", subset[0].min(), subset[0].max())
````
